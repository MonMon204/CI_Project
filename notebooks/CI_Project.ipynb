```python
import numpy as np
import sys
import os

project_path = r"C:\Users\Mariam\Desktop\CI_Project"
sys.path.append(project_path)

import importlib
import lib.layers
importlib.reload(lib.layers)

from lib.network import Sequential
from lib.layers import Dense
from lib.activations import Tanh
from lib.losses import mse_loss, mse_grad

def gradient_check_dense(eps=1e-5):
    X = np.array([[0, 0, 1, 1],
                  [0, 1, 0, 1]], dtype=float)
    Y = np.array([[-1, 1, 1, -1]], dtype=float)

    model = Sequential()
    dense1 = Dense(2, 8)
    model.add(dense1)
    model.add(Tanh())
    dense2 = Dense(8, 1)
    model.add(dense2)
    model.add(Tanh())

    out = model.forward(X)
    grad_loss = mse_grad(Y, out)

    grad = grad_loss
    for layer in reversed(model.layers):
        grad = layer.backward(grad)

    analytical_dW = dense1.dW.copy()

    numerical_dW = np.zeros_like(dense1.W)
    for i in range(dense1.W.shape[0]):
        for j in range(dense1.W.shape[1]):
            old_val = dense1.W[i, j]

            dense1.W[i, j] = old_val + eps
            loss_plus = mse_loss(Y, model.forward(X))

            dense1.W[i, j] = old_val - eps
            loss_minus = mse_loss(Y, model.forward(X))

            numerical_dW[i, j] = (loss_plus - loss_minus) / (2 * eps)
            dense1.W[i, j] = old_val

    diff = np.linalg.norm(analytical_dW - numerical_dW) / (
        np.linalg.norm(analytical_dW) + np.linalg.norm(numerical_dW) + 1e-8
    )
    print("Relative difference:", diff)

gradient_check_dense()
```
